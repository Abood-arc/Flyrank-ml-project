## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
print("""Decision this improves: which pages are likely to decline (or recover) in the near future,
so a content team can act before the drop is already visible in a normal 90-day report.

Who acts, and how: a content editor or SEO strategist, deciding where to spend limited
review/refresh time. Instead of only reacting to pages that have already lost traffic, they'd
get an early-warning list to check proactively.

Cost of a wrong call: two different failure directions, and they cost differently.
A false positive (flagging a healthy page as at-risk) wastes an editors limited review time
on a page that didnt need it. A false negative (missing a page thats about to decline) means
the team finds out the same way they do today — after the traffic is already gone — which is
the exact problem this project is trying to improve on.

Why data/ML helps here specifically: a simple rule like "pages with dropping position are
at risk  is too blunt — normal short-term noise, seasonality, and consolidation (traffic moving
to a sibling page) can all look like decline without actually being decline. Telling those apart
from a real, sustained pattern needs more signals combined than a human can weigh by eye, which
is exactly where a learned model earns its place over a fixed if-statement.""")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/Abood-arc/Flyrank-ml-project"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("../..")  # two levels up: work/notebooks -> work -> repo root

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found — check the path above"
print("Starter data found. You're ready.")
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how common is decline already, in just a 90-day snapshot?
decline_rate = (df["trend_direction"] == "down").mean()
print(f"Share of pages currently trending down: {decline_rate:.1%}")

# Number 2: how many pages are stale AND still getting real traffic. This
# turns out to be tiny (0.1%) — not because visible pages are rare (55.8%
# are), but because staleness itself is rare: only 0.6% of pages have gone
# 180+ days without an update, and the median page was updated just 20 days
# ago. That's the point: clients already keep content fresh, so a rule that
# only flags "old and visible" pages would find almost nothing to act on —
# reinforcing why this lane looks at decline directly, not staleness as a proxy.

# Number 2b: staleness on its own (no visibility condition) -- the two figures
# section 1's framing leans on, computed live here instead of typed in.
stale_180 = (df["days_since_last_update"] >= 180).mean()
median_days_since_update = df["days_since_last_update"].median()
print(f"Share of pages 180+ days without an update (no visibility condition): {stale_180:.1%}")
print(f"Median days since last update: {median_days_since_update:.0f}")

# Number 3: how many pages are old enough to have a real track record —
# this matters because momentum prediction needs prior history to learn from.
old_pages = (df["content_age_days"] >= 365).mean()
print(f"Share of pages older than 1 year (enough history for a trend): {old_pages:.1%}")
```
