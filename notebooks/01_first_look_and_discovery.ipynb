## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

**Rules:** describe what you observe as *observed / directional* — never "I proved Google's algorithm." Keep client data out of anything you publish.

```python
import matplotlib.pyplot as plt

# ============================================================
# Discovery (my turn): does CTR-by-position hold across content types?
# Extends Discovery B by splitting CTR-by-position-tier further by content_type,
# to see which content_type underperforms at the SAME position tier (a fair
# comparison, since raw CTR across different tiers is not comparable).
# ============================================================

# Same filter as Discovery B: keep only pages with enough traffic to trust the
# CTR number (a page with 2 impressions and 1 click has "50% CTR" that means nothing).
visible = df[df["impressions_90d"] >= 100]

# Logical best-to-worst tier order, used for both grids below so rows line up
# and the plot reads left-to-right as "best ranked -> worst ranked."
tier_order = ["top_3", "page_1", "striking", "page_3_5", "deep"]

# --- Grid 1: mean CTR per (position_tier, content_type) ---
# Group by TWO columns at once so we compare content_types WITHIN each tier,
# instead of letting position differences contaminate the comparison.
ctr_grid = visible.groupby(["position_tier", "content_type"])["ctr"].mean()
ctr_grid = ctr_grid.unstack().reindex(tier_order)

# --- Grid 2: sample size (n) behind each cell above ---
# Same shape, but COUNTING rows instead of averaging CTR. A mean from a handful
# of pages is noise wearing a costume - this tells us which cells to trust.
count_grid = visible.groupby(["position_tier", "content_type"]).size()
count_grid = count_grid.unstack().reindex(tier_order)

print("Mean CTR (%) by position tier x content type:")
print(ctr_grid.round(3))
print("\nSample size (n pages) behind each cell above:")
print(count_grid)

# --- Plot ---
ax = ctr_grid.plot(
    kind="bar",
    figsize=(9, 5),
    title="Mean CTR by position tier, split by content type (impressions >= 100)"
)
ax.set_xlabel("Position tier (best to worst)")
ax.set_ylabel("CTR (%)")
ax.legend(title="Content type")
plt.tight_layout()
plt.show()

# ============================================================
# Findings (verified against sample size - see count_grid above)
# ============================================================
# CTR collapses as position tier worsens for all three content types, confirming
# Discovery B's pattern holds even after splitting by content type.
#
# CONFIRMED (adequate sample size, n >= ~40-226 per cell):
# comparison article is consistently the WEAKEST of the three content types at
# every tier with enough data to trust:
#   page_1:    comparison 0.141%  vs feedly 0.905%  vs keyword 0.346%  (n=226/221/8186)
#   striking:  comparison 0.147%  vs feedly 0.358%  vs keyword 0.256%  (n=76/75/5752)
#   page_3_5:  comparison 0.094%  vs feedly 0.166%  vs keyword 0.143%  (n=63/41/5954)
# This is a real, directional pattern across hundreds of pages, not noise.
#
# INSUFFICIENT DATA (do not trust, despite the dramatic-looking number):
# The apparent "feedly article spikes to 2.9% CTR at top_3" is NOT a reliable
# finding - only n=5 feedly pages and n=1 comparison-article page sit behind the
# top_3 row. Per the ~30-50 row sample-size floor, this is noise, not signal.
#
# NO DATA: deep tier has 0 comparison-article pages and only 10 feedly pages -
# too thin to compare content types at that tier at all.
#
# Framing: this is an observed, directional pattern in this 90-day snapshot -
# not a causal claim about WHY comparison articles convert worse (could be
# title/meta style, search intent mismatch, or something else entirely).
```
